# YOLOv8 Object Detection Using Laptop Camera

## Aim
To capture an image using the laptop camera and perform object detection using YOLOv8.

## Step 1: Import Required Libraries

Import OpenCV, Matplotlib, YOLO, and other required libraries.

```python
# Enter your code here
import cv2
import time
import matplotlib.pyplot as plt
from ultralytics import YOLO
```

## Step 2: Load YOLOv8 Model

Load the pretrained YOLOv8 Nano model.

```python
model = YOLO('yolov8n.pt')
```

## Step 3: Open Laptop Camera

Access the laptop camera using OpenCV.

```python
cap = cv2.VideoCapture(0)
```

## Step 4: Capture Image

Wait for 5 seconds and capture one image from the laptop camera.

```python
time.sleep(5)
ret, frame = cap.read()
cv2.imwrite("screen1.jpeg", frame)
```

```python
time.sleep(5)
ret, frame = cap.read()
cv2.imwrite("screen2.jpeg", frame)
```

```python
time.sleep(5)
ret, frame = cap.read()
cv2.imwrite("screen3.jpeg", frame)
```

```python
cap.release()
```

## Step 5: Perform Object Detection

Apply YOLOv8 to the captured image.

```python
results1 = model("screen1.jpeg")
results2 = model("screen2.jpeg")
results3 = model("screen3.jpeg")
```

## Step 6: Display Detected Image

Display the image with bounding boxes and object labels.

```python
result_img1 = results1[0].plot()
plt.imshow(cv2.cvtColor(result_img1, cv2.COLOR_BGR2RGB))
plt.title("Scene 1 Detection")
plt.axis("off")
plt.show()

result_img2 = results2[0].plot()
plt.imshow(cv2.cvtColor(result_img2, cv2.COLOR_BGR2RGB))
plt.title("Scene 2 Detection")
plt.axis("off")
plt.show()

result_img3 = results3[0].plot()
plt.imshow(cv2.cvtColor(result_img3, cv2.COLOR_BGR2RGB))
plt.title("Scene 3 Detection")
plt.axis("off")
plt.show()
```

## Step 7: Release Camera

Release the laptop camera after capturing the image.

```python
cap.release()
```
